```python
import os
from git import Repo
import pandas as pd
import plotly.express as px
from datetime import datetime

# 1) Extrai commits do author
repo_path = '/content/polars'
repo = Repo(repo_path)
top_email = 'ritchie46@gmail.com'

dates = []
for commit in repo.iter_commits():
    if commit.author.email.lower() == top_email.lower():
        dates.append(commit.committed_datetime)

# 2) Monta DataFrame e remover timezone
df = pd.DataFrame({'date': dates})
# Converte para UTC e remove informação de timezone
df['date'] = pd.to_datetime(df['date'], utc=True).dt.tz_convert(None)

# 3) Agrupa por ano-mês
df['year_month'] = df['date'].dt.to_period('M').astype(str)
grouped = df.groupby('year_month').size().reset_index(name='commits')

# 4) Gerar todos os meses até agora
start = pd.Period('2020-05', freq='M')
end   = pd.Period(datetime.today(), freq='M')
all_periods = pd.period_range(start, end, freq='M').astype(str)

full = pd.DataFrame({'year_month': all_periods})
full = full.merge(grouped, on='year_month', how='left').fillna(0)
full['commits'] = full['commits'].astype(int)

# 5) Plotar gráfico de barras
fig = px.bar(
    full,
    x='year_month',
    y='commits',
    color='commits',
    title=f'Atividade Mensal de {top_email}',
    labels={'year_month': 'Ano-Mês', 'commits': 'Número de Commits'}
)
fig.update_layout(xaxis_tickangle=45)
fig.show()
```

```python
import os
from git import Repo
import pandas as pd
import plotly.express as px
from collections import Counter

# 1) aponta para o repositório local
repo_path = '/content/polars'
repo = Repo(repo_path)

# 2) Conta quantas vezes cada arquivo foi alterado
file_changes = Counter()
for commit in repo.iter_commits():
    for f in commit.stats.files.keys():
        file_changes[f] += 1

# 3) Monta DataFrame com as Top 20
df_files = (
    pd.DataFrame(file_changes.items(), columns=['file', 'changes'])
      .sort_values('changes', ascending=False)
      .head(20)
)

# 4) Plota gráfico de barras deitadas (horizontal)
fig = px.bar(
    df_files,
    x='changes',
    y='file',
    color='changes',
    orientation='h',
    title='Top 20 Arquivos Mais Alterados (Churn)',
    labels={'changes': 'Número de Alterações', 'file': 'Arquivo'}
)
fig.update_layout(yaxis={'categoryorder':'total ascending'})
fig.show()
```
