```python
import os
from git import Repo
import pandas as pd
import plotly.express as px
from datetime import datetime

# 1) Extrai commits do author
repo_path = '/content/pandas-ai'
repo = Repo(repo_path)
top_email = 'lele.venturi@gmail.com'

dates = []
for commit in repo.iter_commits():
    if commit.author.email.lower() == top_email.lower():
        dates.append(commit.committed_datetime)

# 2) Monta DataFrame e remover timezone
df = pd.DataFrame({'date': dates})
# Converte para UTC e remove informação de timezone
df['date'] = pd.to_datetime(df['date'], utc=True).dt.tz_convert(None)

# 3) Agrupa por ano-mês
df['year_month'] = df['date'].dt.to_period('M').astype(str)
grouped = df.groupby('year_month').size().reset_index(name='commits')

# 4) Gera todos os meses do inicio até agora
start = pd.Period('2023-04', freq='M')
end   = pd.Period(datetime.today(), freq='M')
all_periods = pd.period_range(start, end, freq='M').astype(str)

full = pd.DataFrame({'year_month': all_periods})
full = full.merge(grouped, on='year_month', how='left').fillna(0)
full['commits'] = full['commits'].astype(int)

# 5) Plota gráfico de barras com Plotly
fig = px.bar(
    full,
    x='year_month',
    y='commits',
    color='commits',
    title=f'Atividade Mensal de {top_email}',
    labels={'year_month': 'Ano-Mês', 'commits': 'Número de Commits'}
)
fig.update_layout(xaxis_tickangle=45)
fig.show()
```

```python
import os
from git import Repo
import pandas as pd
import plotly.express as px
from collections import Counter

# 1) aponta para o seu repositório local
repo_path = '/content/pandas-ai'
repo = Repo(repo_path)

# 2) Conta quantas vezes cada arquivo foi alterado
file_changes = Counter()
for commit in repo.iter_commits():
    for f in commit.stats.files.keys():
        file_changes[f] += 1

# 3) Monta DataFrame com as Top 20
df_files = (
    pd.DataFrame(file_changes.items(), columns=['file', 'changes'])
      .sort_values('changes', ascending=False)
      .head(20)
)

# 4) Plota gráfico de barras deitadas (horizontal)
fig = px.bar(
    df_files,
    x='changes',
    y='file',
    color='changes',
    orientation='h',
    title='Top 20 Arquivos Mais Alterados (Churn)',
    labels={'changes': 'Número de Alterações', 'file': 'Arquivo'}
)
fig.update_layout(yaxis={'categoryorder':'total ascending'})
fig.show()
```

```python
import os
from git import Repo
import pandas as pd
import plotly.express as px


def extract_commit_metrics(repo_path):

    #Extrai contagem de commits por autor e por dia para um repositório Git local.
    #Retorna um DataFrame com colunas: ['repo', 'author', 'date']

    repo = Repo(repo_path)
    data = []
    for commit in repo.iter_commits():
        author = commit.author.name
        date = commit.committed_datetime.date()
        data.append({
            'repo': os.path.basename(repo_path),
            'author': author,
            'date': date
        })
    df = pd.DataFrame(data)
    return df


def aggregate_metrics(df):

    #Agrega contagem de commits por dia para o repositório.
    #Retorna DataFrame com ['repo', 'date', 'commits']

    grouped = df.groupby(['repo', 'date']).size().reset_index(name='commits')
    return grouped


def plot_time_series(df):

    #Plota série temporal de commits por repositório.

    fig = px.line(
        df,
        x='date',
        y='commits',
        color='repo',
        title='Atividade de Commits ao Longo do Tempo'
    )
    fig.update_layout(xaxis_title='Data', yaxis_title='Número de Commits')
    fig.show()


if __name__ == '__main__':
    # Lista de caminhos dos repositórios a serem analisados
    repo_paths = ['/content/pandas-ai', '/content/pytorch', '/content/polars']

    # Extrai e agrega métricas para todos os repositórios
    all_metrics = []
    for path in repo_paths:
        df = extract_commit_metrics(path)
        daily = aggregate_metrics(df)
        all_metrics.append(daily)

    # Junta todos os resultados em um único DataFrame
    combined_metrics = pd.concat(all_metrics, ignore_index=True)

    # Gera gráfico de linha com todos os repositórios
    plot_time_series(combined_metrics)

```
